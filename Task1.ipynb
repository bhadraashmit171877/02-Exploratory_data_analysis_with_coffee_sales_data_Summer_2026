```python
def caesar_cipher(text, shift, mode):
    result = ""

    # Adjust shift for decryption
    if mode == 'decrypt':
        shift = -shift

    for char in text:
        # Encrypt/Decrypt uppercase characters
        if char.isupper():
            result += chr((ord(char) + shift - 65) % 26 + 65)
        # Encrypt/Decrypt lowercase characters
        elif char.islower():
            result += chr((ord(char) + shift - 97) % 26 + 97)
        # Leave numbers, symbols, and spaces unchanged
        else:
            result += char

    return result

def main():
    print("--- Caesar Cipher Tool ---")
    while True:
        choice = input("Do you want to (E)ncrypt, (D)ecrypt, or (Q)uit? ").upper()

        if choice == 'Q':
            print("Goodbye!")
            break
        elif choice in ['E', 'D']:
            mode = 'encrypt' if choice == 'E' else 'decrypt'
            message = input("Enter your message: ")

            try:
                shift = int(input("Enter the shift value (integer): "))
                result = caesar_cipher(message, shift, mode)
                print(f"\nResult ({mode}ed): {result}\n")
            except ValueError:
                print("Error: Shift value must be a whole number.\n")
        else:
            print("Invalid choice. Please enter E, D, or Q.\n")

if __name__ == "__main__":
    main()
```
